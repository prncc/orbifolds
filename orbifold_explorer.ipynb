## Configuration

```python
# --- Matrix A ---
A = build_fermat_matrix([2, 3, 10, 15])  # Diagonal K3 (m=30, weights=(15,10,3,2))
# A = build_fermat_matrix([4, 4, 4, 4])  # Fermat quartic
# A = build_fermat_matrix([5, 5, 5, 5, 5])  # Fermat quintic
# A = matrix([[3,1,0,0],[1,3,0,0],[0,0,3,1],[0,0,1,3]])  # L2 x L2

# --- Symmetry group generators ---
group_generators = [[1] * A.nrows()]

m = get_d(A)
det_A = det(A)
print(f"A = \n{A}")
print(f"\nWeights: {tuple(get_weights(A))}")
print(f"Degree m = {m}")
print(f"det(A) = {det_A}")
```

```python
# Find first prime p with p = 1 (mod m)
p = next(p for p in Primes() if (p - 1) % det_A == 0)
print(f"First eligible prime: p = {p}")

R = Zp(p, 256)
G = generate_symmetry_group(group_generators, A)

all_elements = []
for lam in G:
    all_elements.extend(get_lambda_sector_basis(A, G, lam, m))

hodge_to_elements = group_elements_by_hodge(all_elements)
print(f"|G| = {len(G)}, cohomology elements: {len(all_elements)}")
```

## Cohomology Elements

```python
table(build_cohomology_table(hodge_to_elements, p, R, simplify=True))
```

## Point Counts

```python
N1 = count_points_orbifold(A, m, p, 1, R)
N2 = count_points_orbifold(A, m, p, 2, R)
table([
    ["f", "q = p^f", "N(F_q)"],
    [1, p, N1],
    [2, p^2, N2],
])
```

## Zeta Function

```python
zeta_denom = compute_orbifold_zeta_denominator(all_elements, p, R)
display(Math(render_zeta_latex(zeta_denom, p)))
```

```python
# Point counts from log zeta expansion: N_n = n * [t^n] log Z(t)
num_terms = 10
R_series = PowerSeriesRing(QQ, 't', default_prec=num_terms + 5)
log_zeta = -R_series(zeta_denom).log()

rows = [["n", "N_{F_{p^n}}"]]
for n in range(1, num_terms + 1):
    N_n = Integer(n * log_zeta[n])
    rows.append([n, N_n])
table(rows)
```
